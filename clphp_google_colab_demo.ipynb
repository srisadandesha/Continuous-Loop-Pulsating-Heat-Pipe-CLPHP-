```python
# 4. Run Quickstart Physics Simulation
from clphp.physics.geometry import PHPGeometry
from clphp.physics.fluids import get_fluid
from clphp.simulation.solver import simulate_clphp
from clphp.analysis.plotting import plot_simulation_dashboard
import matplotlib.pyplot as plt

# 5-turn Acetone CLPHP
geo = PHPGeometry(n_turns=5, d_inner=0.002, fill_ratio=0.50)
res = simulate_clphp(geometry=geo, fluid_name='acetone', q_in=60.0, t_span=(0.0, 20.0))

print(f"Simulation Completed!")
print(f"Steady Thermal Resistance: {res.steady_state_r_th:.4f} K/W")
print(f"Dominant Frequency: {res.oscillation_frequency_hz:.2f} Hz")

fig = plot_simulation_dashboard(res)
plt.show()
```
